### 1.What is a Normal Distribution, and what are its key mathematical properties?

```python
A normal distribution is a bell-shaped distribution where most values are around the average,
and fewer values are found as we move away from the average.

-- Key properties:
    
Bell-shaped and symmetric.
Mean = Median = Mode.
Total area under the curve = 1.
About 68% of values fall within 1 standard deviation of the mean.
About 95% fall within 2 standard deviations.
About 99.7% fall within 3 standard deviations.
```

### 2.What techniques do you use to detect and handle outliers in a dataset?

```python
Detection techniques:
-Z-score method: Flag points with |Z| > 3 (or another threshold), 
    assuming roughly normal data. 
-IQR rule: Compute Q1, Q3, then IQR = Q3 − Q1; 
    outliers are below Q1 − 1.5·IQR or above Q3 + 1.5·IQR.
-Visual tools: Box plots, scatter plots, 
    histograms to spot extreme points.
-Model-based / distance methods: Isolation Forest, 
    Local Outlier Factor, Mahalanobis distance (for multivariate data).
    
Handling strategies:
-Investigate first: Check for data entry errors or special causes.
-Remove only if clearly erroneous or if they distort analysis and are not of interest.
-Transform (e.g., log) to reduce skew and the influence of extremes.
-Robust methods: Use median, IQR, or robust regression instead of mean/variance-based methods.
```

### 3.What is a p-value? If your p-value is 0.1, would you reject or fail to reject the null hypothesis (H₀)?

```python
A p-value is the probability, assuming the null hypothesis (H₀) is true,
of getting a test statistic at least as extreme as the one observed.

Common significance level: α = 0.05.

If p = 0.1, then p > 0.05, so you fail to reject H₀ (not enough evidence against it).
```

### 4.	Explain the Central Limit Theorem (CLT) and its significance in data analytics.

```python
The Central Limit Theorem says that when we take sufficiently large random samples from a population, 
the distribution of the sample mean becomes approximately normal, even if the original population is not normal.

For example, even if individual salaries are highly skewed, 
the means of many sufficiently large samples of salaries will tend to form a normal distribution.

Why is it important?

It allows us to:
-Make estimates about populations.
-Construct confidence intervals.
-Perform hypothesis tests.
```

### 5.If a dataset consists of 10 identical values, does it follow a normal distribution? Why or why not? What are its mean and standard deviation?

```python
No, A set of 10 identical values is degenerate: all mass is at one point, 
with zero spread. A normal distribution must have σ > 0 and a bell-shaped curve over a range of values.

Mean = that common value (say c).
Standard deviation = 0 (no variability).

```

### 6.	Define Interquartile Range (IQR), Q₁, and Q₃. How do they relate to box plots?

```python
Q1 = First quartile → 25% of data is below it.
Q3 = Third quartile → 75% of data is below it.
IQR = Q3 − Q1.

The IQR represents the middle 50% of the data.

In a box plot:

Q1 → left/bottom side of the box
Q3 → right/top side of the box
The box length represents the IQR.
The line inside the box represents the median.
Points outside the whiskers may represent outliers.
```

### 8.	What probability sampling techniques are you familiar with?

```python
1. Simple Random Sampling
Everyone has an equal chance of being selected.

2. Stratified Sampling
Divide the population into groups, or strata, and randomly sample from each group.
Example: Divide people by age group and sample from each group.

3. Systematic Sampling
Select every kth person from a list.
Example: Every 10th person.

4. Cluster Sampling
Divide the population into clusters and randomly select some clusters.
Example: Select 10 schools and survey students from those schools.
```

### 9.	Explain what a Confidence Interval is and how to interpret it correctly.

```python
Definition: A range of values derived from sample statistics 
            that is likely to contain the true population parameter.

Interpretation: A 95% confidence interval means that if repeated,samples are taken 
                and intervals computed for each,roughly 95% of those intervals will 
                contain the true population parameter.
```

### 10.	What is the difference between Type I (α) and Type II (β) errors?

```python
Type I Error (α):
We reject H₀ when H₀ is actually true.
-False positive.

Type II Error (β):
We fail to reject H₀ when H₀ is actually false.
-False negative.

Easy example:
Suppose H₀ says "The person is not sick."

Type I → We say sick when they aren't.
Type II → We say not sick when they actually are.
```

### 11.	What are standard deviation and variance?

```python
Both measure how spread out the data is.

Variance:
It is the average squared distance of observations from the mean.

Standard Deviation:
It is the square root of variance.

So:
Standard deviation = √Variance

For example, if variance = 25:
                Standard deviation = 5.
```

### 12.	What is skewness?

```python
Skewness tells us about the asymmetry of a distribution.

Positive/right skew:
Long tail toward the right.

Usually:
Mean > Median
Example: income data.

Negative/left skew:
Long tail toward the left.

Usually:
Mean < Median

Zero skewness:
Approximately symmetric distribution.
```

### 15.	When will you apply following tests?
-	Z-test
-	T-test
-	Chi-Square
-	ANOVA

```python
Z-Test:
Use a Z-test mainly when:
You are testing a population mean/proportion.
Population standard deviation is known, especially for a mean test.
Or you have a sufficiently large sample under appropriate assumptions.

Example:
"I want to test whether the average weight is 70 kg when the population standard deviation is known."

-----------------------------------------------------------------------------------------------------------
T-Test:
Use a T-test when comparing means,
particularly when population standard deviation is unknown.

Common types:
One-sample t-test → Compare one sample mean with a known/reference value.
Independent t-test → Compare means of two independent groups.
Paired t-test → Compare the same subjects before and after something.

Example:
"I want to compare the average salary of employees in two different departments."

--------------------------------------------------------------------------------------------------------------
Chi-Square Test:
Use a Chi-Square test mainly for categorical data.

Common uses:
Test of independence
Goodness-of-fit
Example:
"I want to check whether gender and product preference are related."

Both variables are categorical.
--------------------------------------------------------------------------------------------------------------

ANOVA:
ANOVA = Analysis of Variance
Use it when comparing the means of 3 or more groups.

Example:
"I want to compare the average salary of employees from 3 different departments."

Instead of doing multiple t-tests, we can use ANOVA to test whether there is
evidence that at least one group mean differs.
```
