## Titanic Case study

```python
# replace 'male' with 1 and 'female' with 0
1. using userdefined function + apply
2. using lambda + apply
3. using map + dictionary
4. using map + lambda
5. using replace
```

```python
import pandas as pd

df=pd.read_csv(r"C:\Users\shubh\Downloads\Data\Data\Titanic.csv")
df.head()
```

```python
df["Gender"].unique()
```

#### 1. using userdefined function + apply

```python
def userdef(n):
    if n=="male":
        return 1
    else:
        return 0
```

```python
df["Gender1"]=df["Gender"].apply(userdef)
df.head()
```

#### 2.using lambda + apply

```python
df["Gender2"]=df["Gender"].apply(lambda x: 1 if x=="male" else 0)
df.head()
```

#### 3. using map + dictionary

```python
df["Gender3"]=df["Gender"].map({"male":1,
                               "female":0})
df.head()
```

#### 4. using map + lambda


```python
df["Gender4"]=df["Gender"].map(lambda x:1 if x=="male" else 0)
df.head()
```

#### 5. using replace

```python
df["Gender5"]=df["Gender"].replace(['male','female'],[1,0])
df.head()
```
